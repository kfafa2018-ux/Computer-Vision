Why Did the Original Request Fail? (A Lesson in Web Etiquette & HTTP 403)
If you ran a basic requests.get(url) and received a <Response [403]> or None, congratulations—you just hit your first real-world web scraping hurdle!

When Python's requests library contacts a server, it automatically introduces itself with a default "User-Agent" header: python-requests/x.y.z.

Major websites like Wikipedia actively block requests that use this generic Python signature. They do this to protect their servers from poorly written bot loops, aggressive scrapers, and automated attacks. The server sees an unidentified script knocking on the door and responds with HTTP 403: Forbidden.

To fix this, we need to be good web citizens and provide a custom User-Agent header. This tells the server who is asking for the data and why.

Take a look at the updated cell below: we pass a headers dictionary containing a basic identifier along with our request.

```python
import requests
from bs4 import BeautifulSoup

# Wikipedia requires a descriptive User-Agent header identifying the client
headers = {
    'User-Agent': 'StudentEducationalScraper/1.0 (classroom training; contact@school.edu)'
}

#base_url = 'https://en.wikipedia.org/wiki/Jupiter'

#updating url
base_url = 'https://www.amazon.com/'

# Pass headers into the request
response = requests.get(base_url, headers=headers)

print(f"Status Code: {response.status_code}")

if response.status_code == 200:
    soup = BeautifulSoup(response.text, 'html.parser')
    print("Page Title:", soup.title.string)
else:
    print("Failed to retrieve the page.")
```

##### Let us now look using google for downloading images

The working of the code here will be simple
We put in the required link in which the crawler scraps for data

After finding the required data, we use shutil to download the required amount of images we need

```python
import os
import requests
from bs4 import BeautifulSoup

def download_wikimedia_images(query, quantity=5):
    # Wikimedia Commons search endpoint
    search_url = f"https://commons.wikimedia.org/w/index.php?search={query}&title=Special:MediaSearch&type=image"

    headers = {
        'User-Agent': 'EducationalScraper/1.0 (ClassroomLab; student@school.edu)'
    }

    print(f"Fetching from: {search_url}")
    response = requests.get(search_url, headers=headers)

    if response.status_code != 200:
        print(f"Request failed with status code {response.status_code}")
        return

    soup = BeautifulSoup(response.text, 'html.parser')

    # Grab all img tags inside the search result grid
    img_tags = soup.find_all('img')
    print(f"Found {len(img_tags)} total <img> tags on page.")

    os.makedirs(query, exist_ok=True)
    count = 0

    for img in img_tags:
        if count >= quantity:
            break

        src = img.get('src')
        if not src:
            continue

        # Handle protocol-relative URLs (//upload.wikimedia.org/...)
        if src.startswith('//'):
            src = 'https:' + src

        # Filter out UI icons/badges, keeping actual media files
        if 'upload.wikimedia.org' not in src:
            continue

        try:
            img_data = requests.get(src, headers=headers, timeout=5).content
            filepath = os.path.join(query, f"image_{count + 1}.jpg")
            with open(filepath, 'wb') as f:
                f.write(img_data)
            count += 1
            print(f"Downloaded ({count}/{quantity}): {filepath}")
        except Exception as e:
            print(f"Failed to fetch image: {e}")

    print(f"\nDone. Successfully saved {count} images to folder '{query}/'.")

download_wikimedia_images("mars rover", quantity=5)
```

## Scraping Product Features from Amazon

To extract specific details like the product image, brand, price, and customer rating from an Amazon product page, we'll follow these steps:
1. Define the Amazon product URL.
2. Make a request to the URL with a proper `User-Agent` to avoid being blocked.
3. Parse the HTML content using `BeautifulSoup`.
4. Use CSS selectors or tag inspection to locate and extract the desired information.

```python
import requests
from bs4 import BeautifulSoup

# IMPORTANT: Replace this with the actual Amazon product URL you want to scrape
product_url = 'https://a.co/d/0a1VZ3Yu' # Example product URL

headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}

print(f"Attempting to scrape: {product_url}")
response = requests.get(product_url, headers=headers)

if response.status_code == 200:
    soup = BeautifulSoup(response.content, 'html.parser')

    # 1. Product Image
    product_image_url = 'Not Found'
    # Try common ID first
    product_image_tag = soup.find('img', {'id': 'landingImage'})
    if product_image_tag and 'src' in product_image_tag.attrs:
        product_image_url = product_image_tag['src']
    elif product_image_tag and 'data-a-image-src' in product_image_tag.attrs: # Sometimes it's in data-a-image-src
        product_image_url = product_image_tag['data-a-image-src']
    else:
        # Fallback: look for other potential image attributes or tags in common locations
        main_image_div = soup.find('div', {'id': 'imgTagWrapperId'}) # Common wrapper
        if main_image_div:
            fallback_img = main_image_div.find('img')
            if fallback_img and 'src' in fallback_img.attrs:
                product_image_url = fallback_img['src']

    # 2. Brand
    brand = 'Not Found'
    # Try common ID first
    brand_tag = soup.find('a', {'id': 'bylineInfo'})
    if brand_tag:
        brand_text = brand_tag.text.strip()
        # Clean up text if it contains 'Visit the ... Store'
        brand = brand_text.replace('Visit the ', '').replace(' Store', '')
    else:
        # Fallback: look for span with brand info, often near the title
        brand_span = soup.find('div', {'id': 'bylineInfo_feature_div'})
        if brand_span:
            brand_link = brand_span.find('a', class_='a-link-normal')
            if brand_link:
                brand = brand_link.text.strip()
            else:
                # Sometimes brand is directly in a span without a link
                brand_span_text = brand_span.find('span', class_='a-color-secondary')
                if brand_span_text and 'Brand:' in brand_span_text.text:
                    brand = brand_span_text.text.replace('Brand:', '').strip()

    # 3. Price
    price = 'Not Found'
    price_whole = soup.find('span', class_='a-price-whole')
    price_fraction = soup.find('span', class_='a-price-fraction')
    price_symbol = soup.find('span', class_='a-price-symbol')

    if price_whole and price_fraction and price_symbol:
        price = f"{price_symbol.text.strip()}{price_whole.text.strip()}{price_fraction.text.strip()}"
    else:
        # Fallback for other price formats, e.g., if it's a range or special offer
        price_offscreen = soup.find('span', class_='a-offscreen') # Often contains full price for screen readers
        if price_offscreen:
            price = price_offscreen.text.strip()
        else:
            # Try finding price in a div with price-related class
            price_display = soup.find('span', class_='a-price a-text-price a-size-medium apexPriceToPay')
            if price_display:
                price_val = price_display.find('span', class_='a-offscreen')
                if price_val:
                    price = price_val.text.strip()

    # 4. Customer Rating
    customer_rating = 'Not Found'
    customer_rating_tag = soup.find('span', class_='a-icon-alt')
    if customer_rating_tag:
        customer_rating = customer_rating_tag.text.strip()
    else:
        # Fallback: sometimes rating is in a different span
        rating_span = soup.find('span', class_='a-size-base a-color-secondary')
        if rating_span and 'out of 5 stars' in rating_span.text:
            customer_rating = rating_span.text.strip()

    print("\n--- Product Details ---")
    print(f"Product Image URL: {product_image_url}")
    print(f"Brand: {brand}")
    print(f"Price: {price}")
    print(f"Customer Rating: {customer_rating}")

else:
    print(f"Failed to retrieve the Amazon page. Status code: {response.status_code}")
    print("Please check the URL or try a different User-Agent if the issue persists.")
```
