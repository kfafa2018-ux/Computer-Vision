## Import Libraries needed

```python
import tensorflow
from tensorflow import keras
```

## Implement AlexNet Architecture

```python

model=tensorflow.keras.models.Sequential([
    keras.layers.Conv2D(filters=128, kernel_size=(11,11), strides=(4,4), activation='relu', input_shape=(64,64,3)),
    keras.layers.BatchNormalization(),
    keras.layers.MaxPool2D(pool_size=(2,2)),
    keras.layers.Conv2D(filters=256, kernel_size=(5,5), strides=(1,1), activation='relu', padding="same"),
    keras.layers.BatchNormalization(),
    keras.layers.MaxPool2D(pool_size=(3,3)),
    keras.layers.Conv2D(filters=256, kernel_size=(3,3), strides=(1,1), activation='relu', padding="same"),
    keras.layers.BatchNormalization(),
    keras.layers.Conv2D(filters=256, kernel_size=(1,1), strides=(1,1), activation='relu', padding="same"),
    keras.layers.BatchNormalization(),
    keras.layers.Conv2D(filters=256, kernel_size=(1,1), strides=(1,1), activation='relu', padding="same"),
    keras.layers.BatchNormalization(),
    keras.layers.MaxPool2D(pool_size=(2,2)),
    keras.layers.Flatten(),
    keras.layers.Dense(1024,activation='relu'),
    keras.layers.Dropout(0.5),
    keras.layers.Dense(1024,activation='relu'),
    keras.layers.Dropout(0.5),
    keras.layers.Dense(10,activation='softmax')


])
```

## Tabular summary of The model

```python
model.summary()
```

## Visualizing the Model Architecture

```python
from tensorflow.keras.utils import plot_model

# Visualize the model architecture
plot_model(model, to_file='alexnet_structure.png', show_shapes=True, show_layer_names=True)
```

## Beginner-Friendly Guide to Your AlexNet Notebook

Welcome! This notebook implements **AlexNet**, which is a landmark convolutional neural network (CNN) that won the ImageNet competition in 2012 and sparked the modern deep learning boom. Below is a breakdown of what the code does and what the output diagrams/tables mean.

---

### 1. The Code: What are we building?
We are using **TensorFlow** and **Keras** to build a `Sequential` model (meaning layers are stacked linearly, one after another).

* **`Conv2D` (Convolutional Layer):** These act like scanners. They pass small filters (like searchlights) over the image to detect features like edges, corners, and complex textures.
* **`BatchNormalization`:** This normalizes the inputs to each layer during training. It helps the network learn faster and makes it much more stable.
* **`MaxPool2D` (Pooling Layer):** These layers reduce the spatial size (width & height) of the image representation, keeping only the most important features. This prevents overfitting and speeds up processing.
* **`Flatten`:** Turns the 2D spatial feature maps into a long 1D list of numbers so they can be passed to normal dense neural network layers.
* **`Dense` (Fully Connected Layer):** Standard neural network layers where every neuron is connected to every neuron in the previous layer. They do the final classification decisions.
* **`Dropout`:** Temporarily "turns off" a random half (0.5 or 50%) of the neurons during training. This prevents the model from relying too heavily on specific nodes, forcing it to learn more robust patterns.

---

### 2. Understanding your Outputs

#### A. The Model Summary Table (`model.summary()`)
When you ran `model.summary()`, Keras produced a table listing each layer, its "Output Shape", and "Param #":
* **Output Shape:** Shows how the dimensions of your image data change. For example, your input of `(64, 64, 3)` (a 64x64 pixel image with 3 color channels: Red, Green, Blue) gets progressively smaller in height/width but deeper in "filters" (from 3 channels to 128, then 256 filters) as it goes deeper.
* **Param # (Parameters):** This is the number of weights and biases (the "brain connections") that the computer has to train and learn.

#### B. The Model Visualization (`plot_model`)
The visual chart generated and saved as `alexnet_structure.png` is a flowchart representing this exact data flow. Each box represents one of the layers, showing what goes in (**input**) and what comes out (**output**) to help you verify that your data sizes are transforming correctly from start to finish.
